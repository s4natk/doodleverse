# What the bitmap files actually contain

Each Quick, Draw! category file is a `uint8` array shaped `(N, 784)`. One row is one drawing, already fit into a 28×28 square. This notebook measures that square. It does not train a model.

A drawing made in the app is a list of pen points. Those points are not centered in a 28×28 frame. The network will learn the raster style stored here unless a later step maps a live stroke into the same numeric picture.

```python
from pathlib import Path
from doodleverse.data.explore import explore

report = explore(
    Path("../../data/raw"),
    Path("../../docs/figures"),
    Path("../../docs/experiments.md"),
)
report["ink_polarity"], round(report["mean_pixel"], 2)
```

```python
import numpy as np
from doodleverse.preprocess.bitmap import normalize_quickdraw_bitmap

stored = np.load(Path("../../data/raw/sun.npy"), mmap_mode="r")
row = np.asarray(stored[0])
tensor = normalize_quickdraw_bitmap(row, report["ink_polarity"])
tensor.shape, tensor.dtype, float(tensor.min()), float(tensor.max())
```
